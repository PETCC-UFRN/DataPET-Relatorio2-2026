Este notebook tem o objetivo de reunir e tratar os dados sobre o aumento da temperatura global.


### Request dos dados sobre a média da temperatura anual global 

```python
import time
from pathlib import Path

import numpy as np
import pandas as pd
import requests

URL = "https://data.giss.nasa.gov/gistemp/graphs/graph_data/Global_Mean_Estimates_based_on_Land_and_Ocean_Data/graph.txt"
DATA = Path("dados_tratados/dados_analise_climatica")
DATA.mkdir(exist_ok=True)

def getData(URL):
    return requests.get(URL).text

text = getData(URL=URL)
colunas = []
linhas = []
count = 1
for line in text.splitlines():
    if(count == 4):
        colunas = line.split()
    elif(count > 5):
        linhas.append(line.split())
    count += 1
    
print(colunas)
print(linhas)
```

### Montagem do dataframe pandas
os dados fornecidos pela api não apresentam nenhum erro de formatação ou valores faltantes

```python
df = pd.DataFrame(linhas, columns= colunas).astype(float)
df.head()
```

### Download dos dados


```python
df.to_csv(Path("dados_tratados/dados_analise_climatica/temperatura_global.csv"), index=False)
```
